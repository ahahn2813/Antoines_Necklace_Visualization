```python
#import class
from Antoines_NecklaceClass import AntoineNecklace
```

## Instantiation

We construct an instance with the settings $14$ in level 1, $14$ in level 2, and $10$ in level 3.

```python
necklace = AntoineNecklace(N_l1=14, N_l2=14, N_l3=10)
```

## Level 0

```python
necklace.mesh_res = 100 #increase resolution for higher accuracy
necklace.file_name = "level_0.png"
necklace.eye = [1.5,1.5,1.8]
necklace.generate_level_zero(parent_l0_C=5) #generate torus with radius set to 5
```

## Level 1

```python
necklace.mesh_res = 50 # reduce resolution to decrease compute time
necklace.file_name = "level_1_14.png"
necklace.generate_level_one(parent_l1_C=15) #generate level 1 with large torus radius at 15
```

## Level 2

```python
necklace.mesh_res = 20 # reduce resolution to decrease compute time
necklace.file_name = "level_2_14_and_14.png"
necklace.generate_level_two(parent_l2_C=25)
```

## Level 3

```python
necklace.N_l1 = 10 #reduce number of tori in each level 1 to decrease compute time and memory pressure
necklace.N_l2 = 10 #reduce number of tori in each level 2 to decrease compute time and memory pressure
necklace.mesh_res = 10 # reduce resolution to decrease compute time and memory pressure
necklace.file_name = "level_3_10_10_and_10.png"
necklace.generate_level_three(parent_l3_C=30)
```
